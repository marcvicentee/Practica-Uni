```python
#GRUPO E
#Emmanuel Kwabena Akenuwa Afful
#Edgar Guix Insa
#Marc Linares Carreño
#Marc Vicente Graell

#Libreria para usar el math.
import math

#Funciones de caclulos aritmeticos basicos, los numeros se reciben en una tupla, mediante *numeros. Despues la recorremos con el for para hacer las operaciones
def suma(*numeros):
  totalsum = 0

  for n in numeros:
    totalsum += n

  return totalsum

def resta(*numeros):
  totalres = numeros[0]

  for n in numeros[1:]:
    totalres -= n

  return totalres

def multiplicacion(*numeros):
  totalmul = 1

  for n in numeros:
    totalmul *= n

  return totalmul

def division(*numeros):
  totaldiv = numeros[0]

  for n in numeros[1:]:

    #comprobamos que el divisor no sea 0
    if n == 0:
      return "No se puede dividir entre 0"

    totaldiv /= n

  return totaldiv


#Funcion para el calculo de notas medias ponderadas, comprobamos q no nos den notas y pesos validos, despues agrupamos cada elemento de notas, con su correspondiente en pesos, y devolvemos el total
def nota_pond(notas, pesos):
  suma_notas = 0
  suma_pesos = 0

  if len(notas) != len(pesos):
    return "Nota invalida, cantidad de notas y pesos distintas"

  if sum(pesos) == 0:
    return "La suma de pesos no puede ser 0"

  for nota, peso in zip(notas, pesos):
    suma_notas += nota * peso
    suma_pesos += peso

  return suma_notas / suma_pesos


#Funciones para la conversion entre bases numericas, decimal, binario, hexadecimal

def dec_bin(numerobin):
  return bin(numerobin)

def dec_hex(numerohex):
  return hex(numerohex)


#Funcion para el calculo de potencias de 2, elevamos el 2 con el numero introducido por parametro
def pot_dos(x):
  return 2 ** x

#Funcion para el calculo de logaritmos en base 2, calculamos el logaritmo con la libreria importada
def log_dos(x):

  #comprobamos q el numero sea positivo antes de calcular el logaritmo
  if x<=0:
    return "No puedes introducir menor o igual que 0"

  return math.log2(x)

#Funciones para la conversion de unidades, introducimos un numero por parametro y lo convertimos
def mib_gib(x):
  return x * 0.00104858

def gib_mib(x):
  return x * 1024

#Inicializamos la comprobacion a 0 y hacemos un while con la condicion en true, para que ejecute el programa almenos 1 vez -->
comprobacion = 0

while comprobacion == 0:

  #Titulo
  print("*Calculadora Multifuncion*\n\n")

  #Lista de opciones para escojer de la calculadora
  print("1.Operaciones artimeticas basicas\n")
  print("2.Calculo de nota media ponderada\n")
  print("3.Conversion entre bases numericas\n")
  print("4.Calculo de potencias de 2\n")
  print("5.Logaritmos en base 2\n")
  print("6.Conversion de unidades (MiB/GiB...)\n")

  #lee una opcion, solo es valido escribir un int
  mensaje = int(input("\nEscoje una opcion de la calculadora: "))

  #opcion 1
  if mensaje == 1:

    #lista de operaciones aritmeticas basicas
    print("\n1.Suma")
    print("2.Resta")
    print("3.Multiplicacion")
    print("4.Division")

    #lee una opcion, solo es valido escribir un int
    opcion = int(input("\nEscoje una opcion: "))

    #opcion suma
    if opcion == 1:

      #divide la operacion con el simbolo + y convierte cada numero a float y luego muestra el resultado accediendo a la funcio de la suma
      operaciones = input("\nIntroduce la suma: ")
      numeros = [float(n) for n in operaciones.split("+")]

      print("\n\nEl resultado de la suma es: ", suma(*numeros))

    #opcion resta
    elif opcion == 2:

      #divide la operacion con el simbolo - y convierte cada numero a float y luego muestra el resultado accediendo a la funcion de la resta
      operaciones = input("\nIntroduce la resta: ")
      numeros = [float(n) for n in operaciones.split("-")]

      print("\n\nEl resultado de la resta es: ", resta(*numeros))

    #opcion multiplicacion
    elif opcion == 3:

      #divide la operacion con el simbolo * y convierte cada numero a float y luego muestra el resultado accediendo a la funcion de la multiplicacion
      operaciones = input("\nIntroduce la multiplicacion: ")
      numeros = [float(n) for n in operaciones.split("*")]

      print("\n\nEl resultado de la multiplicacion es: ", multiplicacion(*numeros))

    #opcion division
    elif opcion == 4:

      #divide la operacion con el simbolo / y convierte cada numero a float y luego muestra el resultado accediendo a la funcion de la division
      operaciones = input("\nIntroduce la division: ")
      numeros = [float(n) for n in operaciones.split("/")]

      print("\n\nEl resultado de la division es: ", division(*numeros))

    #escribir otro numero q no sea: 1,2,3,4, no es valido
    else: print("\nOpcion no valida")

  #opcion 2
  elif mensaje == 2:

    #lee las notas y los pesos, y luego divide cada numero con el simbolo "," y luego muestra el resultado accediendo a la funcion de las notas medias ponderadas
    notas = input("\nIntroduce las notas separadas por comas(6,4,8): ")
    pesos = input("\nIntroduce los pesos separados por comas(20,30,50):")
    notas = [float(n) for n in notas.split(",")]
    pesos = [float(n) for n in pesos.split(",")]

    print("\n\nLa nota media ponderada es: ", nota_pond(notas, pesos))

  #opcion 3
  elif mensaje == 3:

    #lista de opciones de conversion de bases numericas
    print("\n1.Decimal-Binario")
    print("2.Decimal-Hexadecimal")

    #lee las opciones de conversion
    opciones = int(input("\nEscoje una de las opciones: "))

    #opcion 1
    if opciones == 1:

      #lee un decimal y muestra el resultado en binario, accediendo a la funcion dec_bin
      numerobin = int(input("\n\nEscribe un numero decimal entro: "))

      print("\nEl numero en binario es: ", dec_bin(numerobin))

    #opcion 2
    elif opciones == 2:

      #lee un decimal y muestra el resultado en hexadecimal, accediendo a la funcion dec_hex
      numerohex = int(input("\n\nEscribe un numero decimal entero: "))

      print("\nEl numero en Hexadecimal es: ", dec_hex(numerohex))

    #si escoje una opcion que no es 1,2,3,4, la opcion no es valida
    else: print("\nOpcion no valida")


  #opcion 4
  elif mensaje == 4:

    #lee un float y muestra el resultado accediendo a la funcion de la potencia de dos
    x = float(input("\n\nIntroduce el exponente:"))

    print("\nEl resultado de la potencia a 2 es: ", pot_dos(x))

  #opcion 5
  elif mensaje == 5:

    #lee un float y muestra el resultado accediendo a la funcion del logaritmo en base dos
    x = float(input("\nIntroduce el numero del logaritmo: "))

    print("\n\nEl resultado del logaritmo en base 2 es: ", log_dos(x))

  #opcion 6
  elif mensaje ==6:

    #lista de opciones de conversion de unidades
    print("\n1.De MiB a GiB")
    print("2.De GiB a MiB")

    #lee las opciones
    opcion = int(input("\nEscoje una opcion: "))

    #opcion MG
    if opcion == 1:

      #lee un float y muestra el resultado accediendo a la funcion de conversion GiB
      x = float(input("\n\nIntroduce el numero que quieres convertir a GiB: "))

      print("\nEl resultado de la conversion es: ", mib_gib(x), " Gib")

    #opcion GM
    elif opcion == 2:

      #lee un float y muestra el resultado accediendo a la funcion de conversion MiB
      x = float(input("\n\nIntroduce el numer que quieres convertir a MiB: "))

      print("\nEl resultado de la conversion es: ", gib_mib(x), " MiB")

    #si escoje una opcion que no es 1,2, la opcion no es valida
    else: print("\nOpcion no valida")

  #si escoje una opcion que no es 1,2,3,4,5,6, la opcion no es valida
  else: print("\nOpcion no valida")


  #Lee un caracter, para continuar dentro del while, o salir
  exit = input("Pulse para continuar, sino pulse 2: ")

  #Si pulsan 2, la comprobacion es diferente de 0, y por lo tanto sale del while, sino se queda
  if exit == "2":

    comprobacion = 1




















```
